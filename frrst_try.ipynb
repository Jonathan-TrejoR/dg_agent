# Langchain

```python
import os
# from langchain_google_vertexai import ChatVertexAI
from langchain_google_genai import ChatGoogleGenerativeAI 
from langchain_core.prompts import ChatPromptTemplate

# 1. Configura el entorno local con los datos de tu proyecto en Liverpool
# (Usa el ID del proyecto que vimos en tus errores de SSH)
PROJECT_ID = "crp-pro-cx-orquestacion"  
LOCATION = "us-central1" # "us-east4"  # Cambia por la región de tus servicios (ej. us-central1, us-east4)

print("🤖 Inicializando conexión local con Gemini Enterprise...")

# 2. Creamos la instancia del modelo de Google en LangChain
# Usamos gemini-1.5-flash que es ultra rápido y barato para pruebas
llm = ChatGoogleGenerativeAI( # con chat vertex ai se usa model_name cuando se selecciona el modelo
    model="gemini-2.5-flash", # Modelos disponibles 2.5 flash y pro (gemini-2.5-flash, gemini-2.5-pro)
    project=PROJECT_ID,
    location=LOCATION,
    temperature=0.2
)

# 3. Creamos una estructura simple de chat (Prompt -> Modelo)
prompt = ChatPromptTemplate.from_messages([
    ("system", "Eres un experto científico de datos senior especializado en Marketing Mix Modeling."),
    ("human", "{input}")
])

# Combinamos el prompt con el modelo usando la sintaxis moderna de LangChain (LCEL)
chain = prompt | llm

# 4. Ejecutamos la prueba
try:
    pregunta = "¿Cuáles son las ventajas de usar la librería Meridian de Google para un modelo MMM?"
    print(f"\nPreguntando: {pregunta}\n" + "-"*50)
    
    respuesta = chain.invoke({"input": pregunta})
    
    print("\nRespuesta de Gemini:")
    print(respuesta.content)
    
except Exception as e:
    print(f"\n❌ Hubo un error en la conexión: {e}")
    print("Verifica que hayas ejecutado 'gcloud auth application-default login' correctamente.")
```

# Langgraph

```python
from langchain.tools import tool
from langchain.chat_models import init_chat_model


model = init_chat_model(
    "gemini-2.5-flash",
    temperature=0
)


# Define tools
@tool
def multiply(a: int, b: int) -> int:
    """Multiply `a` and `b`.

    Args:
        a: First int
        b: Second int
    """
    return a * b


@tool
def add(a: int, b: int) -> int:
    """Adds `a` and `b`.

    Args:
        a: First int
        b: Second int
    """
    return a + b


@tool
def divide(a: int, b: int) -> float:
    """Divide `a` and `b`.

    Args:
        a: First int
        b: Second int
    """
    return a / b


# Augment the LLM with tools
tools = [add, multiply, divide]
tools_by_name = {tool.name: tool for tool in tools}
model_with_tools = model.bind_tools(tools)
```

```python
from langgraph.graph import StateGraph, MessagesState, START, END

def mock_llm(state: MessagesState):
    return {"messages": [{"role": "ai", "content": "hello world"}]}

graph = StateGraph(MessagesState)
graph.add_node(mock_llm)
graph.add_edge(START, "mock_llm")
graph.add_edge("mock_llm", END)
graph = graph.compile()

graph.invoke({"messages": [{"role": "user", "content": "hi!"}]})
```

# Minimum Viable Product (MVP)

```python
import os
from typing import TypedDict, Annotated, List
from google.cloud import bigquery
from langchain_core.tools import tool
from langchain_core.messages import BaseMessage, SystemMessage, HumanMessage
from langchain_google_vertexai import ChatVertexAI, ChatGoogleGenerativeAI 
from langgraph.graph import StateGraph, START, END
from langgraph.graph.message import add_messages
from langgraph.prebuilt import ToolNode



# ==========================================
# 1. DEFINICIÓN DE HERRAMIENTAS (TOOLS)
# ==========================================
# # write file
# @tool

# # send email
# @tool

@tool
def get_bq_job_details(full_job_id: str) -> str:
    """
    Obtiene los detalles (SQL real, slots, bytes) de un Job ejecutado en BigQuery.
    El full_job_id debe tener el formato de la consola: 'proyecto:region.job_id'
    """
    try:
        from google.cloud import bigquery
        client = bigquery.Client()
        
        # Parsear el formato "proyecto:region.job_id"
        partes_proyecto = full_job_id.split(':')
        project = partes_proyecto[0]
        
        partes_region = partes_proyecto[1].split('.')
        location = partes_region[0]
        job_id = partes_region[1]
        
        # Ir a buscar el Job a BigQuery
        job = client.get_job(job_id, project=project, location=location)
        
        # Validar que sea un job de consulta (query)
        if not hasattr(job, 'query'):
            return "El Job especificado no es una consulta SQL (podría ser una carga o exportación)."
            
        # Extraer métricas
        query = job.query
        slot_ms = job.slot_millis or 0
        slot_hours = round(slot_ms / (1000 * 60 * 60), 4)
        bytes_processed = job.total_bytes_processed or 0
        gb_processed = round(bytes_processed / (1024**3), 2)
        
        info = f"""
        📊 Detalles del Job: {job_id}
        - Estado: {job.state}
        - GB Procesados: {gb_processed} GB
        - Consumo Real de Slots: {slot_ms} ms ({slot_hours} Slot-Horas)
        
        📝 Código SQL Ejecutado:
        {query}
        """
        return info
        
    except Exception as e:
        return f"Error al extraer los detalles del Job {full_job_id}: {str(e)}"

@tool
def bq_dry_run(sql_query: str) -> str:
    """
    Ejecuta un dry-run de una consulta SQL en BigQuery para estimar los bytes leídos y el costo.
    NO ejecuta la consulta real, solo simula el cobro.
    """
    try:
        # Asegúrate de pasar el project_id si es necesario: bigquery.Client(project="tu-proyecto")
        client = bigquery.Client()
        job_config = bigquery.QueryJobConfig(dry_run=True, use_query_cache=False)
        query_job = client.query(sql_query, job_config=job_config)
        
        bytes_processed = query_job.total_bytes_processed
        tb_processed = bytes_processed / (1024 ** 4)
        cost_usd = tb_processed * 6.25 # Costo aprox $6.25 USD por TB (On-demand)
        
        return f"Dry-run exitoso. Bytes procesados: {bytes_processed} ({tb_processed:.4f} TB). Costo estimado: ${cost_usd:.4f} USD."
    except Exception as e:
        return f"Error al hacer el dry-run en BigQuery: {str(e)}"

@tool
def get_historical_slot_usage(project_id: str, region: str, routine_name: str) -> str:
    """
    Busca en el historial de BigQuery (INFORMATION_SCHEMA.JOBS) el consumo de SLOTS real
    de las últimas ejecuciones de un Stored Procedure o Query.
    
    Args:
        project_id: El ID del proyecto de GCP (ej. 'crp-pro-cx-analitica')
        region: La región de BigQuery (ej. 'region-us' o 'region-us-central1')
        routine_name: El nombre del SP o palabra clave a buscar.
    """
    try:
        client = bigquery.Client()
        
        # Consultamos la vista de JOBS para ver el histórico de los últimos 5 días
        query = f"""
        SELECT 
            job_id,
            DATE(creation_time) as exec_date,
            total_slot_ms,
            ROUND(total_slot_ms / (1000 * 60 * 60), 2) AS slot_hours,
            ROUND(total_bytes_processed / (1024*1024*1024), 2) as gb_processed,
            job_type,
            statement_type
        FROM `{project_id}.{region}.INFORMATION_SCHEMA.JOBS`
        WHERE query LIKE '%{routine_name}%'
          AND creation_time > TIMESTAMP_SUB(CURRENT_TIMESTAMP(), INTERVAL 5 DAY)
          AND state = 'DONE'
        ORDER BY creation_time DESC
        LIMIT 3
        """
        
        query_job = client.query(query)
        resultados = list(query_job.result())
        
        if not resultados:
            return f"No se encontraron ejecuciones recientes para {routine_name} en los últimos 5 días."
            
        info = f"📊 Historial de consumo de SLOTS para '{routine_name}':\n"
        for row in resultados:
            info += (f"- Fecha: {row.exec_date} | Slot-Horas: {row.slot_hours} "
                     f"| Slot-ms: {row.total_slot_ms} | GB Procesados: {row.gb_processed} | Tipo: {row.statement_type}\n")
        
        return info
            
    except Exception as e:
        return f"Error al extraer el historial de slots: {str(e)}"


@tool
def get_table_metadata(table_id: str) -> str:
    """
    Obtiene los metadatos de una tabla en BigQuery (particionamiento y clústeres).
    El table_id debe tener el formato: 'proyecto.dataset.tabla'.
    """
    try:
        client = bigquery.Client()
        table = client.get_table(table_id)
        
        info = f"Tabla: {table_id}\n"
        if table.time_partitioning:
            info += f"- Particionada por: {table.time_partitioning.field} (Tipo: {table.time_partitioning.type_})\n"
        else:
            info += "- Particionamiento: NINGUNO\n"
            
        if table.clustering_fields:
            info += f"- Clústeres: {', '.join(table.clustering_fields)}\n"
        else:
            info += "- Clústeres: NINGUNO\n"
            
        return info
    except Exception as e:
        return f"No se pudo obtener la metadata de la tabla {table_id}. Error: {str(e)}"


# # write file
# @tool

# # send email
# @tool


@tool
def get_stored_procedure_sql(routine_id: str) -> str:
    """
    Extrae el código SQL original de un Stored Procedure directamente de BigQuery.
    El routine_id debe tener el formato: 'proyecto.dataset.nombre_sp'
    """
    try:
        client = bigquery.Client()
        
        # Separar proyecto, dataset y nombre del SP
        partes = routine_id.split('.')
        project = partes[0]
        dataset = partes[1]
        routine_name = partes[2]
        
        # Consultar el INFORMATION_SCHEMA de BigQuery
        query = f"""
        SELECT routine_definition 
        FROM `{project}.{dataset}.INFORMATION_SCHEMA.ROUTINES`
        WHERE routine_name = '{routine_name}'
        """
        
        query_job = client.query(query)
        resultados = list(query_job.result())
        
        if resultados:
            # Retorna el código SQL del Stored Procedure
            return resultados[0].routine_definition
        else:
            return f"No se encontró el Stored Procedure {routine_name}."
            
    except Exception as e:
        return f"Error al extraer el código del Stored Procedure: {str(e)}"



# ==========================================
# 2. CONFIGURACIÓN DEL AGENTE (LANGGRAPH)
# ==========================================

# Definimos el Estado del Grafo (memoria de mensajes)
class AgentState(TypedDict):
    messages: Annotated[list, add_messages]



SYSTEM_PROMPT = """Eres el Agente Experto en Gobierno de Datos (FinOps y DataOps) de la empresa.
Tu objetivo es auditar consultas y Stored Procedures de BigQuery.

Sigue estos pasos:
1. Si el usuario proporciona un nombre de Stored Procedure, usa 'get_stored_procedure_sql' para leer el código.
2. Usa 'get_historical_slot_usage' para ver cuántos SLOTS (cómputo real) consume este proceso históricamente.
3. Usa 'bq_dry_run' en el SQL extraído solo para detectar la cantidad de bytes que lee la lógica.
4. Identifica las tablas y usa 'get_table_metadata' para revisar particiones.

Si el usuario te proporciona un "Job ID" (ej. crp-pro-cx-analitica:US.script_job_...), debes:
1. Usar la herramienta 'get_bq_job_details' para extraer el SQL y los SLOTS reales consumidos por ese job.
2. Extraer los nombres de las tablas del SQL que te devuelve la herramienta.
3. Usar 'get_table_metadata' en esas tablas para ver si están particionadas.
4. Analizar el código SQL para buscar antipatrones y dar sugerencias de optimización basadas en los slots que consumió ese job.

Tu reporte final debe incluir:
- **Consumo de Cómputo (FinOps):** Menciona las Slot-Horas consumidas históricamente. Advierte si el proceso quema demasiados slots.
- **Análisis de Particiones y Antipatrones:** (Mismas instrucciones que antes).
- **Sugerencia:** Si usa demasiados slots, sugiere moverlo a Dataproc/PySpark.
"""




# Instanciamos el LLM (Gemini 1.5 Pro es excelente para tool calling)
llm = ChatGoogleGenerativeAI(model="gemini-2.5-pro", temperature=0)
tools = [
    bq_dry_run, 
    get_table_metadata, 
    get_stored_procedure_sql, 
    get_historical_slot_usage, 
    get_bq_job_details 
]

llm_with_tools = llm.bind_tools(tools)

# Nodo principal del Agente
def chatbot_node(state: AgentState):
    # Si es el primer mensaje, inyectamos el System Prompt
    messages = state["messages"]
    if not any(isinstance(m, SystemMessage) for m in messages):
        messages = [SystemMessage(content=SYSTEM_PROMPT)] + messages
    
    response = llm_with_tools.invoke(messages)
    return {"messages": [response]}

# Construimos el Grafo
graph_builder = StateGraph(AgentState)
graph_builder.add_node("agent", chatbot_node)
graph_builder.add_node("tools", ToolNode(tools)) # Nodo preconstruido para ejecutar las herramientas

# Definimos la lógica de ruteo condicional
graph_builder.add_edge(START, "agent")
# Si el LLM decide usar una tool, va a 'tools', si no, termina (END)
graph_builder.add_conditional_edges("agent", lambda state: "tools" if state["messages"][-1].tool_calls else END)
graph_builder.add_edge("tools", "agent")

# Compilamos el grafo
finops_agent = graph_builder.compile()

# ==========================================
# 3. EJECUCIÓN (PRUEBA DEL MVP)
# ==========================================
if __name__ == "__main__":
    # Query de prueba (simulando un escenario de un analista)
    test_query = "store_procedure_purchase.sql"
    
    print("Iniciando auditoría de consulta...\n")

    # Tú le dices al agente: "Audita el procedimiento crp-pro-cx-analitica.mus_pro_digital_ecom_tablas_finales.SP_CRO_INSERT_FAC_MDD_WEB_PURCHASE_GAP_TRN"
    # Ejecutamos el agente pasándole la consulta
    # inputs = {"messages": [HumanMessage(content=f"Por favor revisa esta query: \n {test_query}")]}
    # 2. Python lee el contenido del archivo SQL
    ruta_archivo = test_query
    with open(ruta_archivo, 'r', encoding='utf-8') as archivo:
        query_sql = archivo.read()
    inputs = {"messages": [HumanMessage(content=f"Por favor revisa esta query: \n {query_sql}")]}
    
    for event in finops_agent.stream(inputs, stream_mode="values"):
        message = event["messages"][-1]
        # Imprimir solo la respuesta final o llamadas a herramientas
        if message.type == "ai" and not message.tool_calls:
            print(message.content)
        elif message.type == "ai" and message.tool_calls:
             print(f"[⚙️ El agente está llamando a las herramientas: {[t['name'] for t in message.tool_calls]}]")

# Execution 2.0

if __name__ == "__main__":
    job_prueba = "crp-pro-cx-analitica:US.script_job_9937b390b4c9b1f8af1b470fd05b500d_2"
    
    inputs = {"messages": [
        HumanMessage(content=f"Audita este job que falló o fue muy lento hoy en la madrugada: {job_prueba}")
    ]}

	


    for event in finops_agent.stream(inputs, stream_mode="values"):
        message = event["messages"][-1]
        
        if message.type == "ai" and not message.tool_calls:
            print("\n🤖 RESPUESTA FINAL:\n")
            print(message.content)
        elif message.type == "ai" and message.tool_calls:
             print(f"[⚙️ Ejecutando herramienta: {[t['name'] for t in message.tool_calls]}]")

```
